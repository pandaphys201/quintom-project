```python
import numpy as np
from getdist import loadMCSamples, plots
from pathlib import Path
```

```python
# Set Main Chain Dictionary
chains = {}
models = {}
chain_dir = Path(r'/home/theppawan/cosmo-research/quintom-project/chains/quintom')
chain_prefixes = [p.with_suffix('') for p in chain_dir.rglob('*.checkpoint')]

for prefix in chain_prefixes:
    print(f"Loading chain: {prefix}")
    chains[prefix.name] = loadMCSamples(str(prefix), settings={'ignore_rows': 0.3})
```

```python
for sample in chains.values():
    p = sample.getParams()
    abs_delta_qtm = np.abs(p.delta_qtm)

    sample.addDerived(abs_delta_qtm, name='abs_delta_qtm', label=r'|\delta_{\mathrm{qtm}}|')

    print(sample.getName(), sample.getInlineLatex('abs_delta_qtm', limit=1).replace(">","=>").replace("<","=<").split("=", 1)[1])
```
